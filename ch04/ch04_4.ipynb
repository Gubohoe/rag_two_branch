```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_core.prompts import PromptTemplate
from langchain_core.runnables import ConfigurableField
from langchain_openai import ChatOpenAI
```

```python
model = ChatOpenAI(temperature=0, model="gpt-4o")
model.invoke("대한민국의 수도는 어디야?").__dict__
```

```python
model = ChatOpenAI(temperature=0).configurable_fields(
    model_name = ConfigurableField(
        id = "gpt_version",
        name = "Version of GPT",
        description = "Official model name of GPTs. ex) gpt-4o, gpt-4o-mini"
    )
)
```

```python
model.invoke(
    "대한민국의 수도는 어디야?",
    config = {"configurable": {"gpt_version": "gpt-3.5-turbo"}},
).__dict__
```

```python
model.with_config(configurable={"gpt_version": "gpt-4o-mini"}).invoke(
    "대한민국의 수도는 어디야?"
).__dict__
```

```python
prompt = PromptTemplate.from_template("{x}보다 큰 위의 난수를 선택합니다.")
chain = (
    prompt | model
)

chain.invoke({"x": 0}).__dict__
```

```python
chain.with_config(
    configurable = {"gpt_version": "gpt-4o"}).invoke({"x": 0 }
).__dict__
```

```python
from langsmith import Client

client = Client()

prompt = client.pull_prompt("teddynote/rag-prompt-korean",dangerously_pull_public_prompt=True).configurable_alternatives(
    ConfigurableField(
        id = "hub_commit",
        name = "Hub Commit",
        description = "Korean RAG prompt by teddynote",
    ),
    default_key = "rag_korean",
    simple = client.pull_prompt("teddynote/simple-summary-korean",dangerously_pull_public_prompt=True),
)
prompt
```

```python
prompt.invoke({"question": "Hello", "context": "World"}).messages
```

```python
prompt.with_config(
    configurable = {"hub_commit": "simple"}
).invoke({"context": "Hello"})
```
